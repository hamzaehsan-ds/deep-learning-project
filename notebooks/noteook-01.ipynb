```python
import keras
import tensorflow as tf
```

```python
import numpy as np
```

```python
#data
xs = np.array(object=[-1,0,1, 2, 3, 4], dtype=np.float32)
ys = np.array(object=[-3, -1, 1, 3, 5, 7], dtype=np.float32)  
```

```python
model = keras.Sequential()
model.add(layer=keras.Input(shape=(1,)))
model.add(layer=keras.layers.Dense(units=1))
```

```python
model.compile(
    optimizer=keras.optimizers.SGD(learning_rate=0.1), 
    loss='mean_squared_error',
    metrics=[keras.metrics.MeanSquaredError()]
)
model.summary()
```

```python
history = model.fit(
    x=xs,
    y=ys,
    epochs=500,
    verbose=1
)
```

```python
y_pred = model.predict(np.array([[10]]))
print(y_pred)
```
